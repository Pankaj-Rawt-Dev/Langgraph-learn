```python
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, START, END
from dotenv import load_dotenv
from typing import TypedDict
```

```python
load_dotenv()
model = ChatOpenAI()
```

```python
class BlogState(TypedDict):
    title:str
    outline:str
    content:str
    rate:float
```

```python
def create_outline(state=BlogState)->StateGraph:
    title = state['title']

    prompt = f'generate a detail outline for the blog on the title- {title}'

    outline = model.invoke(prompt)
    state['outline']=outline
    return state
```

```python
def create_blog(state=BlogState)->BlogState:

    title=state['title']
    outline=state['outline']

    prompt = f'wriete a detail blog on the title- {title}, using the following outline - {outline}'
    content=model.invoke(prompt).content
    state['content']=content
    return state
```

```python
def generate_evaluation(state:BlogState)->BlogState:
    outline=state['outline']
    content=state['content']

    prompt=f'based on the outline-{outline}, please rate my content-{content} in integer format'

    rate=model.invoke(prompt).content
    state['rate']=rate
```

```python
graph=StateGraph(BlogState)

graph.add_node('create_outline', create_outline)
graph.add_node('create_blog', create_blog)
graph.add_node('generate_evaluation','generate_evaluation')

graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog','generate_evaluation')
graph.add_edge('generate_evaluation',END)
workflow=graph.compile()
```

```python
initial_state={'title':'rise of AI in india'}

final_state=workflow.invoke(initial_state)
print(final_state['content'])
```
